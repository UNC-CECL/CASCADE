```python
import numpy as np

storm_path = (
    "/Users/rsahrae/PycharmProjects/PeaIsland_Hindcast/CASCADE/"
    "data/PeaIsland_init/Storms/storms_1992_2010.npy"
)

# Load storm file
storm = np.load(storm_path)

print("Before:")
print(storm[:10])

# CASCADE storm columns:
# 0 = time/year index
# 1 = Rhigh
# 2 = Rlow
# 3 = Period
# 4 = duration

storm[:, 1] = storm[:, 1] * 10   # Rhigh
storm[:, 2] = storm[:, 2] * 10   # Rlow

print("\nAfter multiplying Rhigh and Rlow by 10:")
print(storm[:10])

# Save as a NEW file so you do not overwrite the original
out_path = (
    "/Users/rsahrae/PycharmProjects/PeaIsland_Hindcast/CASCADE/"
    "data/PeaIsland_init/Storms/storms_1992_2010_Rhigh_Rlow_x10.npy"
)

np.save(out_path, storm)

print(f"\nSaved multiplied storm file to:\n{out_path}")
```

```python
import os
import re

directory_path = "/Users/rsahrae/PycharmProjects/PeaIsland_Hindcast/CASCADE/data/1996"

for filename in os.listdir(directory_path):
    match = re.search(r'resample_domain_(\d+)_1996', filename)
    if match:
        index = int(match.group(1))
        if 80 <= index <= 121:
            old_path = os.path.join(directory_path, filename)
            new_filename = f"resampled_domain_{index}_1996.npy"
            new_path = os.path.join(directory_path, new_filename)

            os.rename(old_path, new_path)
            print(f"Renamed '{filename}' → '{new_filename}'")
```
